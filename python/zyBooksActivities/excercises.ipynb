### 10.21.1 - Explain the difference between using a computer program and programming a computer

Using a computer program is using a program or algorithm already created by someone. It is usually an applicaiton that has been developed. Programming a computer is developing the operating system itself, or programming the files the computer needs to run.

## 10.21.2 - Which parts of a computer can store program code? Which can store user data?

Program Code: RAM, storage (SSD or HDD), cpu  
User Data: Storage (SSD or HDD)

### 10.21.3 - Which parts of a computer serve to give information to the user? Which parts take user input?

Give information: Monitors or displays, speakers, headphones, printers, etc.  
Take User Input: Keyboards, Mice, Microphones, etc.
